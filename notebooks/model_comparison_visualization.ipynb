```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set(style="whitegrid")
DATA_PATH = Path('..') / 'model_comparison_1767485622.csv'
# If notebook runs from notebooks/, adjust path to project root
if not DATA_PATH.exists():
    DATA_PATH = Path('..') / 'artifacts' / 'model_comparison_1767485622.csv'
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
# Limpar e preparar dados para visualização
df_clean = df.copy()
# garantir colunas numéricas onde existam métricas vazias
for c in ['accuracy','precision','recall','f1','roc_auc']:
    if c in df_clean.columns:
        df_clean[c] = pd.to_numeric(df_clean[c], errors='coerce')
df_clean['model_label'] = df_clean['run_name'].fillna(df_clean['model']).fillna('unknown')
df_clean.sort_values('f1', ascending=False, inplace=True)
df_clean[['model_label','accuracy','precision','recall','f1','roc_auc']]
```
